```python
from datetime import datetime
from sqlalchemy import Column, Integer, String, create_engine
from sqlalchemy.orm import declarative_base, sessionmaker

# 1. Configuration de la base SQLite locale (fichier créé dans le dossier courant)
DATABASE_URL = "sqlite:///air_quality_local.db"
engine = create_engine(DATABASE_URL, connect_args={"check_same_thread": False})
SessionLocal = sessionmaker(autocommit=False, autoflush=False, bind=engine)
Base = declarative_base()

# 2. Modèle de la table 'users'
class User(Base):
    __tablename__ = "users"

    id = Column(Integer, primary_key=True, autoincrement=True)
    nom = Column(String(50), nullable=False)
    prenom = Column(String(50), nullable=False)
    email = Column(String(120), nullable=False)
    numero_capteur = Column(String(50), nullable=False)

    def __repr__(self):
        return f"<User id={self.id} nom='{self.nom}' prenom='{self.prenom}' capteur='{self.numero_capteur}'>"

# 3. Création des tables dans le fichier SQLite
Base.metadata.create_all(bind=engine)
print("✅ Base SQLite créée : air_quality_local.db")

# 4. Insertion d'un utilisateur test et vérification
with SessionLocal() as session:
    if session.query(User).count() == 0:
        utilisateur_test = User(
            nom="Panier",
            prenom="Léo",
            email="leo.panier55@gmail.com",
            numero_capteur="14990683"
        )
        session.add(utilisateur_test)
        session.commit()
        print("👤 Utilisateur test ajouté !")

    # Lecture de contrôle
    inscrits = session.query(User).all()
    print("Contenu de la table :", inscrits)
```
