```python
import sqlite3

# Connexion directe au fichier SQLite
conn = sqlite3.connect("air_quality_local.db")
cursor = conn.cursor()

# Requête SQL avec syntaxe UPSERT (Insertion ou Mise à jour si conflit sur l'ID)
sql_query = """
INSERT INTO users (id, nom, prenom, email, numero_capteur)
VALUES (1, 'Panier', 'Léo', 'leo.panier55@gmail.com', '14990683')
ON CONFLICT(id) DO UPDATE SET
    email = excluded.email;
"""

cursor.execute(sql_query)
conn.commit()

# Vérification du résultat
cursor.execute("SELECT id, prenom, nom, email, numero_capteur FROM users WHERE id = 1;")
row = cursor.fetchone()
print(f"Enregistrement SQL : ID={row[0]} | Nom={row[1]} {row[2]} | Email={row[3]} | Capteur={row[4]}")

conn.close()
```
